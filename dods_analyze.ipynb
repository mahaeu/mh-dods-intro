# Data Analyzer Intro

Import the necessary packages

````python
from dods.analyze import DataAnalyzer
import pandas as pd
````

## 1) Load the generated CSV into a DataFrame

Load your Dataframe via pandas to your notebook

````python
import pandas as pd
df = pd.read_csv('data/template.csv')
df.head()
````

## 2) Create an analyzer instance

`DataAnalyzer(df)` is the entry point for your analysis workflow. It will make an analyzis 
of your dataset and give you the possibility to use methods to gain general insights.

````python
dfa = DataAnalyzer(df)
````

## 3) Quick summary

`summary()` should give you a compact overview (columns, missingness, types, etc.).

````python
dfa.summary()
````

## 4) Distribution analysis (optional plots)

This inspects distributions of numerical features.

Notes:
- `plots=True` enables plotting.
- `limit=None` means “no limit” (analyze everything).
- If you want the defaults, you can just call:

  ```python
  dfa.analyze_distributions()
  ```

Further options on the other methods mentioned at the end of the output of
```python
dfa.summary()
```

````python
dfa.analyze_distributions(plots=True, limit=None)
````

## 5) Categorical Column analysis (optional plots)

This inspects all the categorical features of your dataset

Notes:
- `plots=True` enables plotting.
- `limit=None` means “no limit” (analyze everything).
- If you want the defaults, you can just call:

  ```python
  dfa.analyze_categoricals()
  ```

Further options on the other methods mentioned at the end of the output of
```python
dfa.summary()
```

````python
dfa.analyze_categoricals()
````

## 6) Relation Analysis

This is one of the more powerful analysis tool. it gives you the possibility to see which features relate to which features. 
The Pearson group on the later part shows which features are similar and are highly correlated


  ```python
  dfa.analyze_relations()
  ```

Further options on the other methods mentioned at the end of the output of
```python
dfa.summary()
```

````python
dfa.analyze_relations()
````
