Import the necessary package

```python
import dods.datagen as dg
```

## 1) Create a template

This creates a default “template” definition that `dg.run(...)` can execute.
If you already have your own template files, you can skip this.
There are same basic templates available. To create them again, you have to set overwrite=True. It's a safety to not delete prior created templates by accident :)
Let's see it in accion.

```python
# Get list of available templates
dg.list_templates()
```

```python
# The default option is "template", if you want to run it again, set overwrite=True
dg.create_template()
# A longer and more descriptive template
dg.create_template('template_long', path=None, overwrite=True)
```

#### Success ✅

You should now have **template.yaml** and **template_long.yaml** created.

Take a moment to read through both files. They may look a bit large at first — that’s intentional.  
They don’t just show the minimum setup, but also demonstrate **what’s possible** and **which patterns work reliably**.

With these templates you can:

- Define columns (numeric & categorical)
- Derive columns from each other (expressions / dependencies)
- Add noise (for more realistic data)
- Clip values (min/max bounds)
- Simulate missing values and outliers (postprocessing)
- Adjust the amount of data by changing "n:"

The best part:  
You can quickly generate new datasets by copy-pasting the contents of **template_long.yaml** into ChatGPT and describing the dataset you want. Afterwards, you simply fine-tune the YAML (columns, distributions, rules, limits) to fit your needs.


## 3) Generate data from the template

This creates a CSV file from your template. By default, it will be generated in:

- `data/template.csv`  
or  
- `data/template_long.csv`

Simply set the path to the **YAML** template file you want to use.


```python
dg.run('template')
# And a second dataset from template_long
dg.run('template_long')
```
