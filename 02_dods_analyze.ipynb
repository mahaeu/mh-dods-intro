## 6) Relation Analysis

This is one of the most powerful analysis tools. It shows **which features relate to each other**, **how strong the dependency is**, and—most importantly—whether you have **redundant** or **isolated** columns.

### What to pay attention to (high impact)

- **Redundancy Groups (Pearson / MI)**  
  These are *clusters of features that carry very similar information*.  
  Example from the output: `inc` and `spend` form a Pearson group → they are highly correlated.  
  ✅ Practical takeaway: depending on your goal, you can often **keep only one** feature from such a group to reduce noise, multicollinearity, and complexity.

- **Isolated Features**  
  These are columns that have **little to no relationship** with the rest of the dataset.  
  They are often:
  - IDs / index-like columns
  - timestamps
  - unique categories
  - independent dimensions  
  ✅ Practical takeaway: isolated features are not “bad”, but they often **don’t help predictive models** unless they represent meaningful structure (e.g., time).

### What the output includes

- **Data Summary** (rows/columns, feature types, missing values, memory usage)
- **Relation Summary** (how many groups, how many isolated features, average |r|, strong-relation density)
- **Dominant (Pearson r)** and **Dominant (Mutual Information)**  
  → shows which features have the strongest connections overall
- **Correlated Feature Pairs**  
  → ranked list of strongest relations, including `Pearson`, `Spearman`, and `MI`, plus relationship type & flags
- **Global Heatmaps**  
  → quick visual overview of correlation and mutual information across all features

Run it like this:

````python
dfa.analyze_relations()
````
