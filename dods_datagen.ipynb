# DODS Starter Notebook (DataGen + Analyze)

This notebook is a **minimal, runnable starter** that shows the typical flow:

1. Import the two packages you use most: `dods.datagen` and `dods.analyze`
2. Create a template dataset definition
3. Generate a CSV from that template
4. Load the CSV into a pandas DataFrame
5. Run analysis via `DataAnalyzer`

> Keep this notebook as your “quick start”. Put deeper explanations and project docs into a README later.

## 0) Environment assumptions

- You have a Python environment with:
  - `dods` installed (your internal package)
  - `pandas` installed
- You run the notebook from a working directory where `dods.datagen` writes output to `data/`
  (e.g. `data/template.csv`).

If paths differ in your setup, adjust the `pd.read_csv(...)` line later.

## 1) Imports

- `dods.datagen` is used to generate datasets from templates.
- `dods.analyze` contains analysis helpers (here: `DataAnalyzer`).

```python
import dods.datagen as dg
```

## 2) Create a template

This creates a default “template” definition that `dg.run(...)` can execute.
If you already have your own template files, you can skip this.
There are same basic templates available. To create them again, you have to set overwrite=True to evade **FileExistError**
Let's see them and create them.

```python
dg.list_templates()
```

```python
# The default option is "template"
#dg.create_template()
# A longer and more descriptive option
dg.create_template('template_long', path=None, overwrite=True)
```

## 3) Generate data from the template

This should create a CSV at something like:

- `data/template.csv`

Set the path to the **yaml** file of the template.
Take some time to read through the created 
- template (short explanation/quick start)
- template_long (in depth explanation of yaml file configs)

```python
dg.run('template')
# And a second dataset from template_long
dg.run('template_long')
```
